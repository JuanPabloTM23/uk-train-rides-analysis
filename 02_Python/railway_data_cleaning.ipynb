#### **3. Limpiar**

##### **3.1 Reemplazar los nulos en "Railcard"**

```python
df['Railcard'] = df['Railcard'].fillna('No Railcard')

df.head()
```

##### **3.2 Conversión de fechas y horas**

```python
# Fecha
df['Date of Purchase'] = pd.to_datetime(df['Date of Purchase'], format='%Y-%m-%d')
df['Date of Journey'] = pd.to_datetime(df['Date of Journey'], format='%Y-%m-%d')

# Hora
columnas_hora = ['Time of Purchase', 'Departure Time', 'Arrival Time', 'Actual Arrival Time']

for col in columnas_hora:
    df[col] = pd.to_datetime(
        df[col],
        format='%H:%M:%S',
        errors='coerce'
    )

df.info()
```

##### **3.3 Agrupar variables equivalentes**

```python
df['Reason for Delay'] = df['Reason for Delay'].replace(
    ['Signal failure', 'Staff Shortage', 'Weather Conditions'], 
    ['Signal Failure', 'Staffing', 'Weather']
)

df['Reason for Delay'].value_counts(dropna=False)
```

#### **4. Nuevas variables**

```python
# 1.Ruta
df['Route'] = df['Departure Station'].str.strip() + ' → ' + df['Arrival Destination'].str.strip()

# 2.Hora de salida
df['Departure Hour'] = df['Departure Time'].dt.hour

# 3.Franja horaria
df['Time Slot'] = pd.cut(
    df['Departure Hour'],
    bins=[0, 6, 12, 14, 18, 24],
    labels=[
        'Early Morning',
        'Morning',
        'Midday',
        'Afternoon',
        'Night'
    ],
    right=False
)

# 4. Día de la semana
df['Day of Week'] = df['Date of Journey'].dt.day_name()

# 5. Duración programada
# 5.1 Crear fecha y hora programadas de salida y llegada
fecha_viaje = df['Date of Journey'].dt.strftime('%Y-%m-%d')

df['Scheduled Departure'] = pd.to_datetime(fecha_viaje + ' ' + df['Departure Time'].dt.strftime('%H:%M:%S'), errors='coerce')

df['Scheduled Arrival'] = pd.to_datetime(fecha_viaje + ' ' + df['Arrival Time'].dt.strftime('%H:%M:%S'), errors='coerce')

# 5.2 Si la llegada programada ocurre después de medianoche
mask = df['Scheduled Arrival'] < df['Scheduled Departure']

df.loc[mask, 'Scheduled Arrival'] += pd.Timedelta(days=1)

# 5.3 Crear fecha y hora de llegada real
df['Actual Arrival'] = pd.to_datetime(
    fecha_viaje + ' ' +
    df['Actual Arrival Time'].dt.strftime('%H:%M:%S'),
    errors='coerce'
)

# 5.4 Ajustar llegadas reales que ocurren después de medianoche
mask_actual = (df['Actual Arrival'].notna() & (df['Actual Arrival'] < df['Scheduled Departure']))

df.loc[mask_actual, 'Actual Arrival'] += pd.Timedelta(days=1)

# 5.5 Duración programada del viaje, en minutos
df['Scheduled Journey Minutes'] = ((df['Scheduled Arrival'] - df['Scheduled Departure']).dt.total_seconds() / 60)

# 6. Minutos de retraso
df['Delay Minutes'] = ((df['Actual Arrival'] - df['Scheduled Arrival']).dt.total_seconds() / 60)

# 7. Anticipación de compra
df['Booking Lead Time'] = (df['Date of Journey'] - df['Date of Purchase']).dt.days

df.head()
```

#### **5. Exportar**
